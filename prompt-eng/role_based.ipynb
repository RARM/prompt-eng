# Role Based Prompting

```python
from clients import bootstrap_client_and_model
from models import ModelOptions

#### (1) Adjust the inbounding prompt (simulating user/system request)
MESSAGE = "Basecamp Knowledge Retrieval Bot"

#### (2) Role-based prompting
ZERO_SHOT_PROMPT = f"""
You are playing the role of three different personas: a System Architect, a Product Manager, and a User. Your task is to generate a structured requirement analysis for the following bot:
**{MESSAGE}**

### Role 1: **System Architect**
You should focus on:
- The technical architecture and data management considerations.
- Scalability and efficiency.
- Integration with external tools and APIs.
- Database design and query optimization.

### Role 2: **Product Manager**
You should focus on:
- Understanding user needs and priorities.
- The core functionality and user interface.
- User access management and permission levels.
- Key features like task tracking and history retrieval.

### Role 3: **User**
You should focus on:
- The ease of use and user experience.
- How the bot handles daily interactions and multi-user usage.
- The types of tasks users can perform and how intuitive it is.
- The responsiveness and speed of the bot.

The analysis should include functional and non-functional requirements from each perspective. Please ensure the responses are comprehensive, focused on each role, and easy to understand.
"""

PROMPT = ZERO_SHOT_PROMPT
```

```python
#### (3) Configure the Model request (simulating Workflow Orchestration)
client, model = bootstrap_client_and_model("llama3.2:latest")

options = ModelOptions(
    seed = 42,
    # temperature=0.7, # Adjusted temperature for more factual output
    # context_window_size=300, # Increased context to accommodate more detailed responses
    # num_predict=300   # Increased number of tokens to generate a more complete response
)

#### (4) Send request to the client with prompt, selected model, and options.
time, response = client.chat_completion(
    message = PROMPT,
    model = model,
    options = options
)
```

```python
print("------ Response ------")
print(response)

if time: print(f'Time taken: {time}ms')
```
