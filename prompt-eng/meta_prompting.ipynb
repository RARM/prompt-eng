### META PROMPTING


```python
from clients import bootstrap_client_and_model
from models import ModelOptions

SYSTEM_PROMPT = f"""
You are an expert AI in software requirement analysis. Your task is to generate a **structured requirement analysis** for the project the user names.  
 

### Step 1: Understand the Task  
Before generating a response, reflect on the key aspects of a requirement analysis:
- The purpose of the bot and its role in improving workflow efficiency.
- The core features required for intelligent information extraction and retrieval.
- Performance, security, and usability considerations.

### Step 2: Generate a Structured Response  
Ensure your response includes the following sections:  

#### 1️ Objective  
Clearly define the bot's purpose and its impact.  

#### 2️ Functional Requirements  
List the core features that enable intelligent knowledge extraction and retrieval.  

#### 3️ Non-Functional Requirements  
Specify constraints related to scalability, security, and system performance.  

#### 4️ User Interaction & Interface  
Explain how users will interact with the bot for seamless knowledge access.  

#### 5️ Data Handling & Security  
Describe how the bot will manage and protect sensitive project data.  

### Step 3: Self-Check Before Submission  
Before finalizing, review your response:
✅ Is it **clear and structured**?  
✅ Does it **cover all critical aspects** of requirement analysis?  
✅ Would a project manager find it **useful and actionable**?  

Now, generate the final structured requirement analysis based on these guidelines.
"""

USER_PROMPT = "Basecamp Knowledge Retrieval Bot"

client, model = bootstrap_client_and_model("llama3.2:latest")


client.set_system_prompt(SYSTEM_PROMPT) 


options = ModelOptions(temperature=0.7)

#### (4) Make the request
time, response = client.chat_completion(
    message=USER_PROMPT,
    model=model,
    options=options
)

print(response)
if time: print(f'Time taken: {time}s')

```
