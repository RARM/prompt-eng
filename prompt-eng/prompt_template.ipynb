# Prompt Template Prompting

Prompt Template Prompting refers to a technique where predefined templates are used to construct effective prompts that guide large language models (LLMs) to generate responses tailored to specific use cases. The templates typically contain static text combined with dynamic input variables, allowing for consistent, reusable, and customizable prompts.

Prompt templates are widely used across various domains, such as:
* **Question Generation**: Templates can generate quiz questions by filling in variables related to topics.
* **Text Summarization**: Static instructions combined with variable documents or inputs allow flexible summarization.
* **Coding Assistance**: Dynamic prompts help LLMs generate code snippets for different programming tasks.

```python
##
## PROMPT TEMPLATE PROMPTING
##

from clients import bootstrap_client_and_model
from models import ModelOptions

#### (1) Adjust the inbounding  prompt, simulating inbounding requests from
####     users or other systems.
####
####     Template prompting would require the user to provide input variables
####     (or fields) that would fill the template. In this case, the template
####     aims to generate a requirement analysis.
####
#### The fields are:
####  - name: The name of the project.
####  - description: A brief description of the project.
####  - budget: The budget allocated for the project.
####  - timeline_info: Information about the project timeline.
####  - risks: Potential project risks (can be an empty list).
####  - features: Key features of the project (can be an empty list).
####  - team_info: Information about the project team (can be an empty list).
project_info = {
    "name": "Huang's Hype Factory",
    "description": ("An AI-powered Discord bot that enhances Basecamp by "
                    "providing knowledge assistance and streamlining "
                    "information access for Digital Resource (DR), a "
                    "marketing agency."),
    "budget": "Not specified.",
    "timeline_info": ("The project roadmap outlines three releases (v0.1, "
                      "v0.2, v0.3) with various functionalities to be "
                      "implemented in each release. It must be completed "
                      "within 2 months."),
    "risks": [
        "Integration challenges with Basecamp APIs.",
        "Accuracy and efficiency of NLP techniques for data extraction",
        "Maintaining user privacy and data security.",
        "Ensuring scalability and performance as data volume grows."
    ],
    "features": [
        "Document upload and storage.",
        "Smart data extraction from Basecamp.",
        "Context-aware search.",
        "Automated summarization of project updates and discussions.",
        "Task and deadline tracking.",
        "Integration with Discord and Basecamp APIs.",
        "Retrieval-augmented generation (RAG) for dynamic responses.",
        "Named entity recognition (NER) for identifying projects, deadlines, "
        "and users.",
        "Semantic search with a vector database."
    ],
    "team_info": [
        "Only 3 engineers are available for the project.",
        "One engineer is experienced in NLP and AI.",
    ]
}

#### (2) Instantiate chatbot client and preferred model.
client, model = bootstrap_client_and_model("llama3.2:latest")

#### (3) Implementing the template prompting technique.
template = """Project Name: {name}
Description: {description}
Project Budget: {budget}
Timeline: {timeline_info}"""

prompt = template.format(**project_info)

# Convert lists to bullet points and add them if they are not empty.
if project_info['risks']:
  prompt += "\n\nPotential Risks:\n"
  prompt += '\n'.join(f"- {risk}" for risk in project_info['risks'])

if project_info['features']:
  prompt += "\n\nKey Features:\n"
  prompt += '\n'.join(f"- {feature}" for feature in project_info['features'])

if project_info['team_info']:
  prompt += "\n\nTeam Information:\n"
  prompt += '\n'.join(f"- {info}" for info in project_info['team_info'])

print("------ Prompt ------")
print(prompt)
```

```python
from IPython.display import Markdown, display

#### (4) Choosing model options.
# options = ModelOptions(temperature=0.7, max_tokens=100)

#### (5) Send request to the client with prompt, selected model, and options.
time, response = client.chat_completion(
    message=prompt,
    model=model,
    #  options=options
)

print("------ Response ------")
display(Markdown(response))

if time: print(f'Time taken: {time}ms')
```
