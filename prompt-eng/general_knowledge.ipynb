```python

from clients import bootstrap_client_and_model
from models import ModelOptions

#### (1) Adjust the inbounding prompt (simulating user/system request)
MESSAGE = "Basecamp Knowledge Retrieval Bot"

#### (2) Generate Knowledge Prompting (Updated Prompt for Knowledge Extraction)
GEN_KNOWLEDGE_PROMPT = f"""
Generate a detailed, structured requirement analysis based on the following knowledge extraction for the bot:  
**{MESSAGE}**  
The bot should efficiently:
- Organize and retrieve project data, extracting relevant details from various project sources.
- Track tasks, deadlines, and milestones in a dynamic, adaptable manner.
- Enable users to access project history, files, communications, and notes in an easily searchable format.
- Provide seamless, context-sensitive search functionality and intelligent data retrieval.
- Support multiple users, ensuring scalability and reliable simultaneous interactions.
- Extract knowledge about user behaviors, project trends, and critical information to help guide decision-making.

Focus on creating a robust, data-driven requirement analysis that prioritizes key functional and non-functional requirements while considering potential knowledge retrieval challenges and optimizations.

Your analysis should also include any necessary insights about data flows, integration points, and key performance metrics.
"""
#### (3) Instantiate ChatBot Client and preferred model
client, model = bootstrap_client_and_model("llama3.2:latest")

#### (4) Choose Model Options
options = ModelOptions(temperature=0.7)

#### (5) Send the request to the client with the prompt, selected model, and options
time, response = client.chat_completion(message=GEN_KNOWLEDGE_PROMPT,
                             model=model,
                             options=options)
print(response)
if time: print(f'Time taken: {time}ms')


```
