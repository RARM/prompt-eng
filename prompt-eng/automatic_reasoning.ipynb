Automatic Reasoning

Zero-shot prompting refers to a technique in prompt engineering where you provide a model with a task without any prior examples. The model is expected to understand and generate a response or complete the task purely based on the given instruction.

In other words, the model is given "zero" prior training examples or demonstrations in the prompt and relies on its pre-trained knowledge to infer what is needed.

## References:
* [Wei et al. (2022)](https://arxiv.org/pdf/2109.01652.pdf): demonstrate how instruction tuning improves zero-shot learning 

```python
from clients import bootstrap_client_and_model
from models import ModelOptions

#### (1) Adjust the inbounding prompt (simulating user/system request)
MESSAGE = "Basecamp Knowledge Retrieval Bot"

#### (2) Instantiate ChatBot Client and preferred model
client, model = bootstrap_client_and_model("llama3.2:latest")

#### (2) Automatic Reasoning Prompt Engineering
AUTO_REASONING_PROMPT = f"""
You are tasked with analyzing the requirements for a bot that will be used in a project management system, called "{MESSAGE}". The bot is expected to assist users with retrieving information, tracking tasks, and providing seamless project-related data. 

Before diving into functional and non-functional requirements, break down the core tasks this bot must perform:
1. Identify how the bot will organize project data. 
2. Determine how the bot can track and manage tasks and deadlines.
3. Understand the features necessary to access project history, files, and communications.
4. Specify the requirements for advanced search and data retrieval.
5. Consider the multi-user interactions that the bot needs to handle, ensuring seamless functionality.

Once you have identified these tasks, proceed to:
- List the key **functional requirements** of the bot.
- List the key **non-functional requirements** of the bot, explaining any underlying reasoning behind each choice.

Please structure your analysis clearly with an emphasis on logical reasoning and steps involved.
"""

PROMPT = AUTO_REASONING_PROMPT

#### (4) Choose Model Options
options = ModelOptions(temperature=0.7)

#### (5) Send the request to the client with the prompt, selected model, and options
time, response = client.chat_completion(message=PROMPT,
                             model=model,
                             options=options)
print(response)
if time: print(f'Time taken: {time}ms')
```
