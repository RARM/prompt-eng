### Prompt Chaining Level 2

```python
##
##  PROMPT CHAINING LEVEL 2
###
from clients import bootstrap_client_and_model
from models import ModelOptions

MESSAGE = "Basecamp Knowledge Retrieval Bot"

#### (2) Zero-Shot Prompt Engineering for Phase 1 - Requirement Gathering
PHASE_1_PROMPT = f"""
Generate a high-level requirement gathering for the following bot:  
**{MESSAGE}**  
The bot should:
- Organize and retrieve project data.
- Track tasks and deadlines.
- Enable users to access project history, files, and communications.
- Provide seamless search functionality and data retrieval.
- Handle multiple users interacting simultaneously.
"""

#### (3) Phase 2: Detailed Requirement Analysis
PHASE_2_PROMPT_TEMPLATE = f"""
Using the output of the previous phase, provide a detailed analysis for each key feature of the {MESSAGE}.  
Provide more specific examples and elaborate on how the bot should handle each of the following features:
- Organize and retrieve project data.
- Track tasks and deadlines.
- Enable users to access project history, files, and communications.
- Provide seamless search functionality and data retrieval.
- Handle multiple users interacting simultaneously.

Previous Output:
__PREVIOUS__OUTPUT__PLACEHOLDER___
"""

#### (4) Phase 3: Final Requirement Breakdown (Functional & Non-Functional)
PHASE_3_PROMPT_TEMPLATE = f"""
Based on the detailed analysis, break down the requirements into clear categories:
1. Functional Requirements
2. Non-Functional Requirements
For each category, list key items that the {MESSAGE} must support. Be detailed and precise in each requirement.

Detailed Analysis:
__PREVIOUS__OUTPUT__PLACEHOLDER___
"""

client, model = bootstrap_client_and_model("llama3.2:latest")
client.set_system_prompt("PROVIDE ONLY WHAT YOU ARE ASKED FOR DO NOT SAY HERES IS X, JUST PROVIDE IT WITHOUT DESCRIBING IT")
options = ModelOptions(temperature=0.7)

time, response = client.chat_completion(
    message=PHASE_1_PROMPT,
    model=model,
    options=options
)

print(response)
if time: print(f'Time taken: {time}s')
```

```python
PHASE_2_PROMPT = PHASE_2_PROMPT_TEMPLATE.replace("__PREVIOUS__OUTPUT__PLACEHOLDER___", response)

time, response = client.chat_completion(
    message=PHASE_2_PROMPT,
    model=model,
    options=options
)

print(response)
if time: print(f'Time taken: {time}s')
```

```python
PHASE_3_PROMPT = PHASE_3_PROMPT_TEMPLATE.replace("__PREVIOUS__OUTPUT__PLACEHOLDER___", response)

time, response = client.chat_completion(
    message=PHASE_3_PROMPT,
    model=model,
    options=options
)

print(response)
if time: print(f'Time taken: {time}s')
```
