## Few Shots

```python

##
## META PROMPTING
###
from clients import bootstrap_client_and_model
from models import ModelOptions

#### (1) Adjust the inbounding prompt (simulating user/system request)
USER_MESSAGE = "Basecamp Knowledge Retrieval Bot"

#### (2) Few-Shot Prompt Engineering
SYSTEM_MESSAGE = \
f"""
Below are structured requirement analyses for different bots. Follow this format to generate a summarized requirement analysis for the users project in their prompt*.

### Example 1: Health Tracking Bot
**Objective:**  
The bot helps users track daily health metrics, including water intake, diet, and weight. It provides personalized advice and reminders.  

**Functional Requirements:**  
- Log daily health metrics via natural language input.  
- Provide personalized health recommendations.  
- Send reminders for water intake and meals.  

**Non-Functional Requirements:**  
- Scalability to support multiple users.  
- Data privacy and encryption for health records.  
- Real-time analysis and tracking.  

**User Interaction & Interface:**  
- Simple commands for logging and retrieving health data.  
- Visual dashboards for progress tracking.  

---

### Example 2: Chemistry Study Companion Bot
**Objective:**  
The bot assists students with chemistry learning by providing interactive Q&A, flashcards, and summaries.  

**Functional Requirements:**  
- Answer chemistry-related questions using natural language processing.  
- Generate flashcards and summaries from uploaded materials.  
- Retrieve relevant study resources.  

**Non-Functional Requirements:**  
- High accuracy in educational content.  
- Efficient handling of large textbooks and notes.  
- User-friendly and engaging study interface.  

**User Interaction & Interface:**  
- Discord-based chatbot interface.  
- Voice-based Q&A for hands-free learning.  

---
Use the same structured format as the examples above.**
"""

#### (2) Instantiate Client and Model
client, model = bootstrap_client_and_model("llama3.2:latest")

#### (3) Adjust the Prompt Engineering Technique to be applied, simulating Workflow Templates
client.set_system_prompt(SYSTEM_MESSAGE) 

#### (4) Set Model Options
options = ModelOptions(temperature=0.7)

#### (4) Make the request
time, response = client.chat_completion(
    message=USER_MESSAGE,
    model=model,
    options=options
)

### YOU DONT NEED TO CONFIGURE ANYTHING ELSE FROM THIS POINT
# Send out to the model
print(response)
if time: print(f'Time taken: {time}s')
```
